```python
df_song = df_anotated[df_anotated['song'] == 'ES_milagro'].reset_index(drop=True)
```

```python
def apply_filter(df):
    df.copy()
    c1 = df['x'] > 0
    c2 = df['y'] < 5

    return df[c1 | c2]
```

```python
# remove  x < 0 & y > 5
df_part = df_song[3256:3718]
df_part_filter = apply_filter(df_part)
df_anotated_filter = apply_filter(df_anotated)
```

```python
def add_cm_for_tag(df_phrase, df_anotated):
    com_df = (df_anotated
              .groupby('phone_base')[['x', 'y']]
              .median()
              .rename(columns={'x': 'com_x', 'y': 'com_y'}))
    
    df_phrase_cm = df_phrase.merge(com_df, left_on='phone_base', right_index=True)

    return df_phrase_cm.drop(columns = ['song'])
```
